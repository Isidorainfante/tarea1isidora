# Pregunta 1
Crea un programa que imprima los números desde -30 hasta 50.

Para cada número, además de imprimirlo, el programa debe mostrar un texto según estas reglas:

si el número es menor que 0, mostrar "Muy frío"
si el número está entre 0 y 24, mostrar "Templado"
si el número es mayor o igual a 25, mostrar "Caluroso"

```python
# lista desde -30 al 50
lista_numeros = range(-30,50)
print(lista_numeros)

for x in range(-30,51):
  if(x < 0):
    print(x, "Muy Frío")
  elif(x >= 0 and x <= 24):
      print(x, "Templado")
  else:
    print(x, "Caluroso")
```

## Pregunta 2

Crea un programa que imprima los números del 1 al 100 que sean divisibles por 5.

Además, el programa debe cumplir estas condiciones:

si el número es divisible por 5, imprimir el número junto con el texto "Bip" si el número también es divisible por 10, imprimir el número junto con el texto "Bip Bip"

```python

#el for i in range(x,y) recorre todos los numeros del x hasta y - 1, por eso se pone 101
for i in range(1, 101):

    #todo numero divisible por 10 es tambien divisible por 5, como el resto es 0 significa que es divisible por 10
    if i % 10 == 0:
        print(f"{i} Bip Bip")

    #sino, si el numero es solo divisible por 5 printeo un solo "bip"
    elif i % 5 == 0:
        print(f"{i} Bip")
```

# Pregunta 3

Crea un programa que solicite al usuario dos números, los sume y luego muestre un mensaje según el resultado.

Reglas

si la suma es menor que 50, mostrar "Nivel principiante"
si la suma es mayor o igual a 50 y menor que 80, mostrar "Nivel intermedio"
si la suma es mayor o igual a 80, mostrar "Nivel pro"
Ejemplo de ejecución

Ingresa primer puntaje: 20
Ingresa segundo puntaje: 25

Resultado: Nivel principiante

```python
#pido los datos al usuario
num1 = int(input("Ingresa primer puntaje: "))
num2 = int(input("Ingresa segundo puntaje: "))

#calculo la suma
suma = num1 + num2

#veo que condiciones cumple
if suma < 50:
    print("Resultado: Nivel principiante")

#uso and para imponer dos condiciones en el mismo if
elif suma >= 50 and suma < 80:
    print("Resultado: Nivel intermedio")

#con la condicion que queda el numero solo puede ser mayor o igual a 80 --> equivale a poner un else
else:
    print("Resultado: Nivel pro")
```

# Pregunta 4

Crea un programa que evalúe el perfil del usuario según su edad y sus gustos.

El usuario debe recibir los siguientes parámetros:

- edad
- le_gusta_musica
- le_gusta_bailar

Y debe retornar uno de los siguientes mensajes:

- "Eres una estrella de la fiesta" si la persona tiene 18 o más años, le gusta la música y le gusta bailar
- "Te gusta la música, pero no tanto bailar" si le gusta la música pero no le gusta bailar
- "Tienes tu propio estilo" en cualquier otro caso

```python
#uso def para definir la funcion
def perfil_musical(edad, le_gusta_musica, le_gusta_bailar): #recibo parametros

    if edad >= 18 and le_gusta_musica == True and le_gusta_bailar == True: #en un mismo if pido 3 condiciones
        return "Eres una estrella de la fiesta"

    elif le_gusta_musica == True and le_gusta_bailar == False: #pido 2 condiciones
        return "Te gusta la música, pero no tanto bailar"

    else: #uso else para cualquier otro caso
        return "Tienes tu propio estilo"

#ejemplo de uso
print(perfil_musical(20, True, True))
print(perfil_musical(25, True, False))
print(perfil_musical(17, False, False))
```
